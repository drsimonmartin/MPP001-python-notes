# Lab learning plan

Over the next few weeks we will cover thes topics listed below which are deisgned to help you learn programming in python with an emphasis on reproducable workflows that are required in engineering and related careers:

0. Basics of python and notebooks
1. Write, execute and debug Python programs.
2. Use variables, functions, conditionals and loops appropriately.
3. Work with arrays and tabular engineering data.
4. Use Python libraries including NumPy, Pandas and Matplotlib.
5. Implement simple engineering calculations and simulations.
6. Developing reproducible computational workflows suitable for MSc-level engineering analysis.

1 to 6 will have assessed exercises associated with them. Together these exercises are worth 10% of the module. 

This assessment will be done by submitting each week's notebook to Gradescope via the appropriate submission point on Learn. Gradescope will evaluate your code, checking that it performs the required functions and check for similarity with the work of others and any other unusual patterns.

## Gen AI use
You should avoid using AI (ChatGPT, Claude, etc.) with the lab problems as **this will significantly interfere** with your learning of programming skills. Some of the notebooks, particularly those towards the end of the course will use Gradescope to embed hidden tests that will prevent GenAI producing useful results. If you have skipped the learning process then you are likely to struggle with solving the notebook problems.
