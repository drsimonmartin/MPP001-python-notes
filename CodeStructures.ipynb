# Lab 2 The building blocks of computer code
In order to get a program with a useful level of complexity there are a few key building blocks. These are common to most computer languages (there will be syntax differences–for example python uses indentation of code to separate blocks whilst many others use curly brackets i.e. `{}`. 

The building blocks key to introductory python programming are described below.

## Conditional statements
These check whether a condition is met and perform different code based on this check. The classic structure is the `if` statement which will execute the following code if the condition tested evaluates to TRUE. `if` statements can be extended in a number of ways: an `else` block adds code that is only run if the condition is not met. In python there is also an `elif` command that can be thought of as 'else if...'

````python

# if example
def Greater(x,y): # returns greater of x or y
    if x > y:
        return x
    elif y > x:
        return y
    else: return x # x&y must be equal, does not matter which is returned

print(Greater(4,5))

````

## Loops
These are used to repeat a block of code either a fixed number of times (usually the `for` loop) or until some condition is met (`while` loop).

### for loops
In most computer languages the for-loop increments/decrements a numeric variable through a range of values. Here is a C example

```c
// Loop in C language - DO NOT copy to python
for (int i = 1; i < 11; ++i)
    // i goes from 1 to 10 in steps of 1
  {
    printf("%d ", i); // print current value of i
  }
```
__Do not__ try and run this code in Jupyter (or any other python system)  Hopefully you can see that it is designed to print out the integers from 1 to 10.

In python the loop goes through a list of values. The must all be the same type, but beyond that there are very few limits. So this example prints out the names of the primary colours:

```python
colours = ['red','green','blue'] # a list of colours
for colour in colours: # start of the loop. The variable colour contains a different item from the list colours each time through the loop (starting with 'red')
    print(f'{colour}')
```

The C example above could be reproduced in python with something like:

```python
for i in range(11): # range produces a list of integers from 0 up to the value sent in the parentheses
    print(i)
```
range() is a function that returns a list of numbers. It defaults to starting at 0, but this can be changed (e.g. `range(1,11)` would give 1, 2, ..., 10).  

::: {.callout-tip}
Loops are often used for calculating sums, products, etc. make sure the variable that you are storing the result in starts from where it is supposed to. For sums (`i = i+NextValue`) this will usually be zero. For products (`y = y*NextValue`) one is a better choice.
:::

### while loops
While loops are another way to repeat a block of code. The loop operates as long as a condition is true. The following example produces the same output as the `for` loop above.

```python
i = 0 # start value
while i < 11: # loop with condition 
    i += 1 # adds 1 to the value of i
    print(f'i = {i}')
    
```
::: {.callout-tip} 
## Check the conditions
It is easy to get the stopping conditions of a while loop wrong and if you're lucky the loop will end sooner than you expect allowing you to see that there is a problem. __However__ it often happens that the stop conditions are never met and the loop attempts to go on until the end of time. Forgetting to increment the loop variable is the most common mistake, but there are plenty of others - be vigilent!

:::


### `break` and `continue`
Python provides functions to alter a loop's operation. These work in both `for` and `while` structures.

`break` stops a loop. The following loop will stop when `i` reaches the value 5:

````python
for i in range(1,8): # range produces a list of integers 
    if i == 5: break
    print(i)
````

__Quick question__ How would you change the loop above so that it printed out the value 5 before stopping?

`continue` goes to the next iteration of the loop rather than stopping the loop entirely:

````python
i = 1 # start value
while i < 9: # loop with condition 
    i += 1 # adds 1 to the value of i
    if i == 5: continue
    print(f'i = {i}')
````

## Functions
We've met these before, but for completeness here is a quick reminder

These are blocks of code that you send values to and the block returns one or more values. `print()` is an example of a built-in function. You can add your own functions to python code either by importing them from another file/script or by defining them in the script you are working on.

A minimal python example would be:

```python
def HelloWorld():
	print('Hello world!')
```
Note the code after the `def` line is indented – this is how python tracks the extent of the current code block. You can see a similar use of indenting in the for loop above.

The brackets after `HelloWorld` can take the names of arguments (parameters) that are to be passed to the function. E.g.

```python
def Quotient(x,y):
	print(x/y)
```

which could be called by `Quotient(2,3)`. Note the order of arguments matters.

You can give arguments default values:

```python
def Sum(x,y=3):
	return(x+y)
```

Calling this with `Sum(4)` will return `7` whereas `Sum(4,2)` will return `6`. Note all arguments after the first with a default value must have default values.
